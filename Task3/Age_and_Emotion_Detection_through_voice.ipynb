```python
# ==========================================
# Task 3: Age and Emotion Detection from Voice
# Internship: NullClass
# Author: Manvitha
# ==========================================

# Install required libraries
!pip install librosa soundfile speechbrain gradio --quiet

# Imports
import librosa
import numpy as np
from speechbrain.pretrained import EncoderClassifier
import gradio as gr

# ==========================================
# Load Pre-trained Emotion Detection Model
# ==========================================
emotion_model = EncoderClassifier.from_hparams(
    source="speechbrain/emotion-recognition-wav2vec2-IEMOCAP"
)

# ==========================================
# Feature Extraction from Audio
# ==========================================
def extract_features(file_path):
    y, sr = librosa.load(file_path, sr=None)
    mfccs = np.mean(librosa.feature.mfcc(y=y, sr=sr, n_mfcc=40).T, axis=0)
    return mfccs

# ==========================================
# Gender and Age Prediction (Dummy Logic)
# Replace with trained models for accuracy
# ==========================================
def predict_gender(features):
    # Example rule: mean < 0 = Male, else Female
    return "Male" if np.mean(features) < 0 else "Female"

def predict_age(features):
    # Example: returns integer between 0–80
    return int(abs(np.mean(features) * 100) % 80)

# ==========================================
# Emotion Prediction for Senior Citizens
# ==========================================
def predict_emotion(file_path):
    out_prob, score, index, text_lab = emotion_model.classify_file(file_path)
    return text_lab[0]

# ==========================================
# Main Logic
# ==========================================
def process_audio(file_path):
    features = extract_features(file_path)
    gender = predict_gender(features)

    if gender == "Female":
        return {"Message": "Upload male voice."}

    age = predict_age(features)

    if age > 60:
        emotion = predict_emotion(file_path)
        return {
            "Gender": gender,
            "Age": age,
            "Senior Citizen": "Yes",
            "Emotion": emotion
        }
    else:
        return {
            "Gender": gender,
            "Age": age,
            "Senior Citizen": "No"
        }

# ==========================================
# GUI with Gradio
# ==========================================
iface = gr.Interface(
    fn=process_audio,
    inputs=gr.Audio(type="filepath", label="Upload Voice (WAV/MP3)"),
    outputs="json",
    title="Age and Emotion Detection from Voice",
    description="Detects age from male voices. If age > 60, also detects emotion."
)

iface.launch()
```
