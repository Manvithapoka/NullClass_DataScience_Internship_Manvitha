```python
# ===============================
# 1️⃣ Install required libraries
# ===============================
!pip install deepface scikit-learn joblib

# ===============================
# 2️⃣ Imports
# ===============================
from deepface import DeepFace
import os
import joblib
from sklearn.linear_model import LogisticRegression

# ===============================
# 3️⃣ Paths
# ===============================
data_dir = "data/nationality"  # <-- Ensure your dataset is uploaded here

# ===============================
# 4️⃣ Prepare Data
# ===============================
race_mapping = {}
nationality_mapping = {}
race_id = 0
nat_id = 0
X, y = [], []

for nationality in os.listdir(data_dir):
    nat_path = os.path.join(data_dir, nationality)
    if not os.path.isdir(nat_path):
        continue

    if nationality not in nationality_mapping:
        nationality_mapping[nationality] = nat_id
        nat_id += 1

    for img_name in os.listdir(nat_path):
        img_path = os.path.join(nat_path, img_name)

        try:
            analysis = DeepFace.analyze(img_path, actions=['race'], enforce_detection=False)
            race = analysis[0]['dominant_race']

            if race not in race_mapping:
                race_mapping[race] = race_id
                race_id += 1

            X.append([race_mapping[race]])
            y.append(nationality_mapping[nationality])
        except Exception as e:
            print(f"⚠️ Skipping {img_name}: {e}")

print("✅ Data preparation done!")
print("Race mapping:", race_mapping)
print("Nationality mapping:", nationality_mapping)

# ===============================
# 5️⃣ Train Model
# ===============================
model = LogisticRegression()
model.fit(X, y)
print("✅ Model trained successfully!")

# ===============================
# 6️⃣ Save Model & Mappings
# ===============================
joblib.dump(model, "nationality_model.pkl")
joblib.dump(race_mapping, "race_mapping.pkl")
joblib.dump(nationality_mapping, "nationality_mapping.pkl")
print("✅ Model and mappings saved successfully!")

# ===============================
# 7️⃣ Download Files
# ===============================
from google.colab import files
files.download("nationality_model.pkl")
files.download("race_mapping.pkl")
files.download("nationality_mapping.pkl")
```
